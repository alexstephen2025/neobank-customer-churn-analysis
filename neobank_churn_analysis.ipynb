```python
import duckdb

query = """
SELECT 
    Geography,
    Gender,
    COUNT(*) AS total_customers,
    SUM(Exited) AS total_churned,
    ROUND(AVG(Exited) * 100, 2) AS group_churn_rate_pct
FROM 'D:/NeoBank_Churn_Project/Churn_Modelling.csv'
GROUP BY Geography, Gender
ORDER BY group_churn_rate_pct DESC;
"""

duckdb.query(query).show()
```

```python
import duckdb

query = """
SELECT 
    CASE 
        WHEN Age < 30 THEN '1. Under 30'
        WHEN Age BETWEEN 30 AND 40 THEN '2. 30-40'
        WHEN Age BETWEEN 41 AND 50 THEN '3. 41-50'
        WHEN Age BETWEEN 51 AND 60 THEN '4. 51-60'
        ELSE '5. Above 60'
    END AS age_group,
    COUNT(*) AS total_customers,
    SUM(Exited) AS total_churned,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM 'D:/NeoBank_Churn_Project/Churn_Modelling.csv'
GROUP BY age_group
ORDER BY age_group;
"""

duckdb.query(query).show()
```
